## Etapa 3. Modelo <a id='model'></a>

### Dividindo os dados de origem em um conjunto de treinamento, um conjunto de validação e um conjunto de teste.

```python
# random_state garante que a divisão seja a mesma a cada execução; stratify mantém a proporção dos planos
df_train_valid, df_test = train_test_split(df, test_size=0.2, random_state=12345, stratify=df['is_ultra'])
df_train, df_valid = train_test_split(df_train_valid, test_size=0.25, random_state=12345, stratify=df_train_valid['is_ultra'])
```

```python
features_train = df_train.drop(['is_ultra'], axis=1)
target_train = df_train['is_ultra']
features_valid = df_valid.drop(['is_ultra'], axis=1)
target_valid = df_valid['is_ultra']
features_test = df_test.drop(['is_ultra'], axis=1)
target_test = df_test['is_ultra']
```

Separamos em 3 os dados: conjunto de treinamento, um conjunto de validação e um conjunto de teste.

### Melhorando a qualidade do modelo

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid_dtc = {
    'max_depth': range(1, 9),
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'criterion': ['gini', 'entropy'],
    'random_state': [5432]
}
```

```python
model_dtc = DecisionTreeClassifier()

# Criando o objeto GridSearchCV
grid_search_dtc = GridSearchCV(estimator=model_dtc, param_grid=param_grid_dtc, cv=5, scoring='accuracy')

# Executando a pesquisa em grade nos dados de treinamento
grid_search_dtc.fit(features_train, target_train)

# Obtendo os melhores parâmetros encontrados
best_params_dtc = grid_search_dtc.best_params_
print("Melhores parâmetros encontrados:", best_params_dtc)

# Obtendo a melhor pontuação de acurácia
best_accuracy_dtc = grid_search_dtc.best_score_
print("Melhor acurácia encontrada:", best_accuracy_dtc)
```

```python
param_grid_rfc = {
    'n_estimators': [50, 100, 200],
    'max_depth': [None, 10, 20],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'max_features': ['sqrt', 'log2', None],
    'random_state': [5432]
}
```

```python
# Criando o modelo RandomForestClassifier
model_rfc = RandomForestClassifier()

# Criando o objeto GridSearchCV
grid_search_rfc = GridSearchCV(estimator=model_rfc, param_grid=param_grid_rfc, cv=5, scoring='accuracy')

# Executando a pesquisa em grade nos dados de treinamento
grid_search_rfc.fit(features_train, target_train)

# Obtendo os melhores parâmetros encontrados
best_params_rfc = grid_search_rfc.best_params_
print("Melhores parâmetros encontrados:", best_params_rfc)

best_accuracy_rfc = grid_search_rfc.best_score_
print("Melhor acurácia encontrada:", best_accuracy_rfc)
```

```python
# Combinações válidas: o solver 'liblinear' aceita as penalidades l1 e l2; o 'lbfgs', só a l2
param_grid_lr = [
    {'solver': ['liblinear'], 'penalty': ['l1', 'l2'], 'C': [0.001, 0.01, 0.1, 1, 10, 100], 'max_iter': [1000], 'random_state': [5432]},
    {'solver': ['lbfgs'], 'penalty': ['l2'], 'C': [0.001, 0.01, 0.1, 1, 10, 100], 'max_iter': [1000], 'random_state': [5432]},
]
```

```python
# Criando o modelo LogisticRegression
model_lr = LogisticRegression()

# Criando o objeto GridSearchCV
grid_search_lr = GridSearchCV(estimator=model_lr, param_grid=param_grid_lr, cv=5, scoring='accuracy')

# Executando a pesquisa em grade nos dados de treinamento
grid_search_lr.fit(features_train, target_train)

# Obtendo os melhores parâmetros encontrados
best_params = grid_search_lr.best_params_
print("Melhores parâmetros encontrados:", best_params)

# Obtendo a melhor pontuação de acurácia
best_accuracy_lr = grid_search_lr.best_score_
print("Melhor acurácia encontrada:", best_accuracy_lr)
```

### Comparação no conjunto de validação

A busca em grade escolheu os hiperparâmetros com validação cruzada dentro do treino. Agora comparo os três modelos no **conjunto de validação**, que eles não viram, e escolho o melhor. Também calculo a acurácia de um modelo ingênuo, que sempre recomenda o plano mais comum (Smart), como **teste de sanidade**: um modelo útil precisa superá-lo.

```python
from sklearn.dummy import DummyClassifier

candidatos = {
    "Árvore de Decisão": grid_search_dtc.best_estimator_,
    "Floresta Aleatória": grid_search_rfc.best_estimator_,
    "Regressão Logística": grid_search_lr.best_estimator_,
}
acuracia_validacao = pd.Series({nome: accuracy_score(target_valid, modelo.predict(features_valid))
                                for nome, modelo in candidatos.items()}).sort_values(ascending=False)
ingenuo = DummyClassifier(strategy="most_frequent").fit(features_train, target_train)
print("Acurácia na validação:")
print(acuracia_validacao.round(3).to_string())
print(f"\nModelo ingênuo (sempre Smart): {accuracy_score(target_valid, ingenuo.predict(features_valid)):.3f}")

nome_escolhido = acuracia_validacao.index[0]
modelo_escolhido = candidatos[nome_escolhido]
```

### Aplicando a melhor algoritmo com os melhores hiperparâmetros para o modelo.

```python
# Fazendo previsões nos dados de teste com o modelo escolhido na validação
predictions_test = modelo_escolhido.predict(features_test)

# Calculando a acurácia das previsões nos dados de teste
accuracy_test = accuracy_score(target_test, predictions_test)
print(f"Modelo: {nome_escolhido}")
print(f"Acurácia do modelo nos dados de teste: {accuracy_test:.3f} (meta: 0,75)")
```

## Conclusão geral <a id='end'></a>

- Com a busca em grade e validação cruzada, a **Floresta Aleatória** teve a melhor acurácia média (0,80), seguida da Árvore de Decisão (0,79) e da Regressão Logística (0,75).
- No **conjunto de validação**, a Floresta Aleatória também foi a melhor (0,82), bem acima do modelo ingênuo que sempre recomenda o plano Smart (0,69), o que confirma que o modelo aprendeu padrões reais.
- No **conjunto de teste**, usado uma única vez, a Floresta Aleatória alcançou **acurácia de 0,82**, acima da meta de 0,75.

**Correções desta versão:** a divisão dos dados passou a ter `random_state` e estratificação (antes, cada execução gerava uma divisão diferente), as células que exibiam os parâmetros de outro modelo foram corrigidas, combinações de hiperparâmetros inválidas foram removidas da busca e o modelo final passou a ser escolhido no conjunto de validação, com um teste de sanidade contra o modelo ingênuo.

Recomenda-se utilizar a Floresta Aleatória para recomendar o plano mais adequado (Smart ou Ultra) aos clientes da Megaline.
